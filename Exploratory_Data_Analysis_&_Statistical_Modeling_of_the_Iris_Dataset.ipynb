# Exploratory Data Analysis & Statistical Modeling of the Iris Dataset

This notebook demonstrates a rigorous, professional approach to Exploratory Data Analysis (EDA) on the classic **Fisher's Iris dataset**. We leverage `pandas`, `numpy`, and `seaborn` to perform data profiling, statistical summarizing, and feature correlation analyses.

```python
import pandas as pd
import numpy as np
import seaborn as sns

# Load a public sample dataset (e.g., Iris)
data = sns.load_dataset('iris')

# --- Step 1: Explore the Data ---
print("--- FIRST 5 ROWS ---")
display(data.head())

print("\n--- DATA INFO ---")
print(data.info())

print("\n--- SUMMARY STATISTICS ---")
display(data.describe())

# --- Step 2: Calculate Basic Statistics ---
numeric_cols = data.select_dtypes(include=[np.number])

print("\n--- DETAILED STATISTICS PER NUMERICAL COLUMN ---")
for col in numeric_cols.columns:
    print(f"\nFeature: {col}")
    print(f"  Mean:               {numeric_cols[col].mean():.4f}")
    print(f"  Median:             {numeric_cols[col].median():.4f}")
    print(f"  Mode:               {numeric_cols[col].mode()[0]:.4f}")
    print(f"  Standard Deviation: {numeric_cols[col].std():.4f}")
    print(f"  Variance:           {numeric_cols[col].var():.4f}")

# Correlation matrix
print("\n--- CORRELATION MATRIX ---")
correlation_matrix = numeric_cols.corr()
display(correlation_matrix)
```
